# Mini-Project: Drowsiness Classification with Machine Learning

**Goal:** Use a small CSV of precomputed eye/behavior measurements to predict the dataset's label.

> **Important limitation:** This is a teaching demo, not a validated driver-safety system. Treat this small dataset as an example until its provenance and labeling method are independently verified. High accuracy here does not prove real-world drowsiness detection.

- **Features (`X`)** are the measurements the model receives.
- **Label (`y`)** is the category the dataset assigns to each row.
- The model learns patterns linking features to labels, then predicts labels for rows it did not train on.

## 1. Import the tools

Pandas loads the CSV, Matplotlib visualizes it, and scikit-learn splits data, trains a model, and evaluates predictions.

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
```

## 2. Load the dataset

Download `dataset_EAR_300.csv` and put it in the **same folder** as this notebook.

Dataset source: [muratgok/drowsiness on GitHub](https://github.com/muratgok/drowsiness)

- `Timestamp_s`: time in seconds when a row was recorded.
- `EAR`: Eye Aspect Ratio, a numerical measure related to eye openness. Lower values can indicate more closed eyes, depending on how it is measured.
- `Blinks_min`: estimated blinks per minute.
- `Label`: the category assigned to that row.

For this exercise, we'll use `EAR` and `Blinks_min` as features. We won't use timestamp as a feature because clock time itself shouldn't be used to guess the state.

```python
df = pd.read_csv("dataset_EAR_300.csv")

print("Dataset shape (rows, columns):", df.shape)
display(df.head())
```

## 3. Understand the data before training

Ask: How many rows and columns are there? What does each row represent? Which columns are measurements and which is the answer? Are there missing values? How many examples do we have for each label?

```python
print("Column names:", df.columns.tolist())
print("\\nData types and non-missing counts:")
df.info()

print("\\nMissing values per column:")
display(df.isna().sum())

print("\\nExamples per label:")
display(df["Label"].value_counts())
```

```python
df["Label"].value_counts().plot(kind="bar")
plt.title("Number of rows in each label")
plt.xlabel("Label")
plt.ylabel("Number of rows")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

### Explain the features in plain language

- **EAR:** Eye Aspect Ratio. Think of it as a numerical signal related to how open the eyes are. It is not a universal drowsiness score on its own.
- **Blinks_min:** Estimated number of blinks per minute. Blink behavior may change with fatigue, but varies between people and situations.
- **Label:** The target category in this dataset (for example, `Awake`, `Fatigue`, `Drowsy`, or `Micro-sleep`). The model learns to predict this category from the measurements.

**Key point:** A feature is an input clue; a label is the answer the model learns to predict. A pattern in a small dataset does not automatically mean the model will work reliably in real life.

## 4. Prepare features (`X`) and labels (`y`)

`X` contains the input columns. `y` contains the category we want to predict.

```python
feature_columns = ["EAR", "Blinks_min"]

X = df[feature_columns]
y = df["Label"]

print("X shape:", X.shape)
print("y shape:", y.shape)
display(X.head())
display(y.head())
```

## 5. Split into training and testing data

- **Training set:** examples the model learns from.
- **Test set:** examples used afterward to check predictions on data it didn't train on.

`stratify=y` helps keep a similar label mix in both sets. This random split is for teaching; because this dataset is small and its collection/labeling process needs verification, the score is not evidence of real-world performance.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

## 6. Train a Decision Tree

A Decision Tree learns a series of if/else-like splits in the features to predict a label. We limit its depth to make it easier to understand and reduce memorization.

```python
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)

print("Model trained!")
```

## 7. Evaluate the model

Accuracy is the fraction of test rows predicted correctly. The classification report shows precision, recall, and F1-score for each label. The confusion matrix shows which labels are mixed up.

Don't focus only on the accuracy number—especially with a small, unverified dataset.

```python
y_pred = model.predict(X_test)

print(f"Test accuracy: {accuracy_score(y_test, y_pred):.2%}")
print("\\nClassification report:")
print(classification_report(y_test, y_pred, zero_division=0))

ConfusionMatrixDisplay.from_predictions(y_test, y_pred, xticks_rotation=35)
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

## 8. Visualize the Decision Tree

Each node splits the data using a feature threshold. The tree may split rows based on EAR or blink rate, then predict a label at the end.

```python
plt.figure(figsize=(16, 8))
plot_tree(model, feature_names=feature_columns, class_names=model.classes_, filled=False, rounded=True)
plt.tight_layout()
plt.show()
```

## 9. Try a prediction

Change the values below to experiment. These are examples only; don't interpret the result as a real diagnosis or a safe-to-drive decision.

```python
example = pd.DataFrame({"EAR": [0.18], "Blinks_min": [26.0]})
prediction = model.predict(example)[0]
print("Predicted dataset label:", prediction)
```
